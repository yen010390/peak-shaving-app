# Peak-Shaving Battery Analysis — run the Streamlit app in Colab

This notebook clones the GitHub repo, installs its dependencies, and launches
the Streamlit app (`app.py`) with a public URL via **localtunnel** — no ngrok
account or token needed.

**Steps:** Runtime → Run all, then open the printed `https://*.loca.lt` link.
The first time it opens, it may ask for a "Tunnel Password" — that's just
your Colab machine's public IP, printed by the second cell below.

```
#@title 1. Clone the repository
#@markdown Replace with your own fork's URL if you changed anything.
REPO_URL = "https://github.com/YOUR-USERNAME/peak-shaving-app.git"  #@param {type:"string"}

import os
if not os.path.isdir("/content/peak-shaving-app"):
    !git clone -q "$REPO_URL" /content/peak-shaving-app
%cd /content/peak-shaving-app
!ls
```

```
#@title 2. Install dependencies
!pip install -q -r requirements.txt
!npm install -q -g localtunnel
```

```
#@title 3. Print the tunnel password (your Colab machine's public IP)
!curl -s https://loca.lt/mytunnelpassword ; echo
```

```
#@title 4. Launch Streamlit + localtunnel, print the public URL
import subprocess, time, re, urllib.request

st_proc = subprocess.Popen(
    ["streamlit", "run", "app.py", "--server.headless", "true", "--server.port", "8501"],
    stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL,
)
time.sleep(8)  # give Streamlit a moment to bind the port

lt_proc = subprocess.Popen(
    ["lt", "--port", "8501"],
    stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True,
)
for _ in range(30):
    line = lt_proc.stdout.readline()
    if "your url is" in line.lower():
        print(line.strip())
        break
print("\nOpen the URL above, enter the tunnel password from step 3 if asked, "
      "and use the app. Re-run this cell if the tunnel drops.")
```

## Stopping

Colab reclaims the runtime automatically when the notebook is closed/idle.
To stop manually, interrupt the kernel (Runtime → Interrupt execution) —
this kills the `streamlit` and `lt` (localtunnel) background processes.
